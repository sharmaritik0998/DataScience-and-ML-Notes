```python
x = np.linspace(0, 4*np.pi, 100)

#Defining Wave 
sin_y = np.sin(x)
cos_y = np.cos(x)

#plotting the wave 
plt.plot(x, sin_y, label = "Sine")
plt.plot(x, cos_y, label = "Cosine")

plt.legend()
plt.title('Sine Wave')
plt.show()
```

```python
x = np.linspace(0, 4*np.pi, 100)

#Defining Wave 
sin_y = np.sin(x)
cos_y = np.cos(x)

#plotting the wave 
plt.plot(x, sin_y, '*')
plt.plot(x, cos_y, '-')

plt.title('Sine Wave')
plt.show()
```

```python
x = np.linspace(0, 4*np.pi, 1000)

y_sin = np.sin(x)
y_cos = np.cos(x)

plt.figure(figsize = (15, 5))

plt.style.use('Solarize_Light2')

plt.plot(x, y_sin, color = 'red')
plt.plot(x, y_cos, color = 'blue')

plt.title('Sine & Cosine Wave')

plt.show()
```

```python
x = np.linspace(0, 4*np.pi, 1000)

y_sin = np.sin(x)
y_cos = np.cos(x)

plt.figure(figsize = (15, 5))

plt.style.use('Solarize_Light2')

plt.plot(x, y_sin, '*', color = 'red', linewidth = 1, label = 'sin')
plt.plot(x, y_cos, '--', color = 'blue', linewidth = 1, label = 'cos')

plt.legend()
plt.title('Sine & Cosine Wave')

plt.show()
```

# Pie Chart

```python
cars = ['AUDI', 'BMW', 'FORD', 'TESLA', 'JAGUAR', 'MERCEDES']
data = [450, 385, 35, 1, 200, 500]

fig = plt.figure(figsize = (7,7))

plt.pie(data, labels = cars)
plt.legend()
plt.show()
```

```python
cars = ['AUDI', 'BMW', 'FORD', 'TESLA', 'JAGUAR', 'MERCEDES']
data = [450, 385, 35, 1, 200, 500]
explode = [.1, .2, .3, .2, .1, .3]

fig = plt.figure(figsize = (7,7))

plt.pie(data, labels = cars, explode = explode)
plt.legend()
plt.show()
```

```python
cars = ['AUDI', 'BMW', 'FORD', 'TESLA', 'JAGUAR', 'MERCEDES']
data = [450, 385, 35, 1, 200, 500]
explode = [0, 0, 0, 0, .1, 0]

fig = plt.figure(figsize = (7,7))

plt.pie(data, labels = cars, explode = explode)
plt.legend()
plt.show()
```

```python
#Adding shadows 
cars = ['AUDI', 'BMW', 'FORD', 'TESLA', 'JAGUAR', 'MERCEDES']
data = [450, 385, 35, 1, 200, 500]
explode = [0, 0, 0, 0, .1, 0]

fig = plt.figure(figsize = (7,7))

plt.pie(data, labels = cars, explode = explode, shadow = True)
plt.legend()
plt.show()
```

```python
#Adding the percentage 
cars = ['AUDI', 'BMW', 'FORD', 'TESLA', 'JAGUAR', 'MERCEDES']
data = [450, 385, 35, 1, 200, 500]
explode = [0, 0, 0, 0, .1, 0]

fig = plt.figure(figsize = (7,7))

plt.pie(data, labels = cars, explode = explode, shadow = True, autopct = '%1.1f%%')
plt.legend()
plt.show()
```

```python
#Adding wedget
#Adding the percentage 
cars = ['AUDI', 'BMW', 'FORD', 'TESLA', 'JAGUAR', 'MERCEDES']
data = [450, 385, 35, 1, 200, 500]
explode = [0, 0, 0, 0, .1, 0]

fig = plt.figure(figsize = (7,7))

plt.pie(data, labels = cars, explode = explode, shadow = True, autopct = '%1.1f%%', wedgeprops= {'edgecolor': 'black', 'linewidth': 2, 'antialiased': True})
plt.legend()
plt.show()
```

```python
#Adding wedget
#Adding the percentage 
cars = ['AUDI', 'BMW', 'FORD', 'TESLA', 'JAGUAR', 'MERCEDES']
data = [450, 385, 35, 1, 200, 500]
explode = [0, 0, 0, 0, .1, 0]

fig = plt.figure(figsize = (7,7))

plt.pie(data, labels = cars, explode = explode, shadow = True, autopct = '%1.1f%%', wedgeprops= {'edgecolor': 'black', 'linewidth': 2, 'antialiased': False})
plt.legend()
plt.show()
```

```python
x = np.linspace(0, 10, 50)
fig , ax = plt.subplots(2,2)
y = np.sin(x)
ax[0,0].plot(x,y)
ax[0,0].set_title('Sin Wave')

y = np.cos(x)
ax[0,1].plot(x,y)
ax[0,1].set_title('Cos Wave')

y = np.tan(x)
ax[1,0].plot(x,y)
ax[1,0].set_title('Tan Wave')

y = np.random.randn(1000)
ax[1,1].hist(y,bins = 40)
ax[1,1].set_title('Random Distribution')

plt.tight_layout()

plt.show()
```

```python
x = np.linspace(0, 10, 50)
y = np.sin(x)

for i in range(1,2):
    fix, (ax1, ax2, ax3) = plt.subplots(1,3,figsize=(18,6))
    ax1.plot(x,y)
    ax2.plot(x,y**2)
    ax3.plot(x,y**3)

    ax1.set_title('Sin Wave')
    ax2.set_title('Sin Wave - Square')
    ax3.set_title('Sin Wave - Cube')

    plt.tight_layout()
    plt.show()
    

# fig , ax = plt.subplots(2,2)

# ax[0,0].plot(x,y)
# ax[0,0].set_title('Sin Wave')

# y = np.cos(x)
# ax[0,1].plot(x,y)
# ax[0,1].set_title('Cos Wave')

# y = np.tan(x)
# ax[1,0].plot(x,y)
# ax[1,0].set_title('Tan Wave')

# y = np.random.randn(1000)
# ax[1,1].hist(y,bins = 40)
# ax[1,1].set_title('Random Distribution')

# plt.tight_layout()

# plt.show()
```
