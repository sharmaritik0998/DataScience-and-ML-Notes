```python
import numpy as np
import pandas as pd 
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv("E:\ML GFG\Data Analysis\googleplaystore.csv")
```

```python
df.head()
```

```python
impute = SimpleImputer(missing_values = np.nan, strategy = 'mean')
impute.fit(df.iloc[:, 2:3].values)
df.iloc[:, 2:3] = impute.transform(df.iloc[:, 2:3].values)

df = df.dropna()
```

```python
df.isnull().sum()
```

```python
df
```

# Q1. How many free apps are there in ART_AND_DESIGN

```python
df = df.values
```

```python
df[1]
```

```python
c = 0
for i in df:
    if i[1] == 'ART_AND_DESIGN' and i[6] == 'Free':
        c += 1

print("There are", c, "Free apps from ART_AND_DESIGN")
```

```python
c = 0
for i in df:
    if i[1] == 'ART_AND_DESIGN' and i[6] == 'Paid':
        c += 1

print("There are", c, "Paid apps from ART_AND_DESIGN")
```

# Q2. How Many Apps are there ART_AND_DESIGN with rating more than 4.5

```python
c = 0
for i in df:
    if i[1] == 'ART_AND_DESIGN' and i[2] > 4.5:
        c += 1

print("There are", c, "Rating is greater than 4.5 apps from ART_AND_DESIGN")
```

# Q3. How many apps are there in FAMILY with rating more than 4.5 and Free?

```python
c = 0
for i in df:
    if (i[1] == 'ART_AND_DESIGN' and i[2] > 4.5 and i[6]=='Free'):
        c += 1

print("There are", c, "Rating is greater than 4.5 apps from ART_AND_DESIGN")
```
