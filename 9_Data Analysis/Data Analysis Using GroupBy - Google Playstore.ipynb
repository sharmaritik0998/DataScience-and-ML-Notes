```python
import numpy as np
import pandas as pd 
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv("E:\ML GFG\Data Analysis\googleplaystore.csv")
```

```python
impute = SimpleImputer(missing_values = np.nan, strategy = 'mean')
impute.fit(df.iloc[:, 2:3].values)
df.iloc[:, 2:3] = impute.transform(df.iloc[:, 2:3].values)

df = df.dropna()
```

# Q1. Name the apps that are in ART_AND_DESIGN with rating more than 4.5 in ascending order WRT Ratings ?

```python
df_pr = df[df['Category']=='ART_AND_DESIGN']
df_pr = df_pr[df_pr['Rating']>4.5]
```

```python
df_pr
```

```python
df_pr.sort_values(by ='Rating', ascending = True)
```

```python
df_pr.sort_values(by ='Rating', ascending = False)
```
